```python
estado = pd.read_csv("perfil_eleitor_secao_2022_SP.csv",
                     usecols = '''NM_MUNICIPIO DS_ESTADO_CIVIL DS_FAIXA_ETARIA DS_GRAU_ESCOLARIDADE QT_ELEITORES_PERFIL QT_ELEITORES_INC_NM_SOCIAL'''.split(),
                     sep = ";", on_bad_lines = 'skip', encoding='Latin 1')
municipio = estado.query('NM_MUNICIPIO == "SÃO JOSÉ DOS CAMPOS" and NM_MUNICIPIO != "-1" and DS_ESTADO_CIVIL != "-1" and DS_FAIXA_ETARIA != "-1" and DS_GRAU_ESCOLARIDADE != "-1" and QT_ELEITORES_PERFIL != "-1" and QT_ELEITORES_INC_NM_SOCIAL != "-1"')
municipio = municipio.drop(columns = 'NM_MUNICIPIO')
```

```python
dataFrame = pd.DataFrame(municipio)
```

```python
dataFrame.sample(25)
```

```python
#Classificação à partir de Estado Civil, Faixa Etária, Escolaridade
estado_civil = dataFrame.groupby('DS_ESTADO_CIVIL')['QT_ELEITORES_PERFIL'].sum().sort_values(ascending=False)
faixa_etaria = dataFrame.groupby('DS_FAIXA_ETARIA')['QT_ELEITORES_PERFIL'].sum().sort_values(ascending=False)
escolaridade = dataFrame.groupby('DS_GRAU_ESCOLARIDADE')['QT_ELEITORES_PERFIL'].sum().sort_values(ascending=False)
```

```python
#Qtd eleitores nome social
nome_social = dataFrame.groupby('QT_ELEITORES_INC_NM_SOCIAL')['QT_ELEITORES_PERFIL'].count()
```
